```python
from pathlib import Path
import json

from src.core.config import DataConfig, OptunaConfig, TrainingConfig
from src.pipelines.train_optuna import run_optuna_time_series_search

def find_project_root():
    cwd = Path.cwd().resolve()
    for p in [cwd, *cwd.parents]:
        if (p / 'src').exists() and (p / 'data').exists():
            return p
    raise RuntimeError('Nie moge znalezc katalogu projektu.')

PROJECT_ROOT = find_project_root()
```

```python
optuna_cfg = OptunaConfig(n_trials=30)
train_cfg = TrainingConfig(epochs=20, early_stopping_patience=5, device="cpu")

auto_summary = run_optuna_time_series_search(
    data_config=DataConfig(),
    optuna_config=optuna_cfg,
    training_config=train_cfg,
    n_splits=5,
)
auto_summary
```

```python
summary_path = PROJECT_ROOT / "reports" / "optuna_summary.json"
summary = json.loads(summary_path.read_text(encoding="utf-8"))
summary
```
