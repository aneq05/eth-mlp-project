```python
from pathlib import Path
import pandas as pd

from src.core.config import DataConfig, SplitConfig, TargetConfig
from src.pipelines.prepare_data import prepare_datasets

def find_project_root():
    cwd = Path.cwd().resolve()
    for p in [cwd, *cwd.parents]:
        if (p / 'src').exists() and (p / 'data').exists():
            return p
    raise RuntimeError('Nie moge znalezc katalogu projektu.')

PROJECT_ROOT = find_project_root()
DATA_PATH = PROJECT_ROOT / "data" / "raw" / "ethusdt_1h.csv"
DATA_PATH
```

```python
raw_df = pd.read_csv(DATA_PATH)
raw_df.head()
```

```python
raw_df.info()
```

```python
ts = pd.to_datetime(raw_df["timestamp"], utc=True, errors="coerce")
print("Rows:", len(raw_df))
print("Min timestamp:", ts.min())
print("Max timestamp:", ts.max())
print("NaN total:", int(raw_df.isna().sum().sum()))
print("Duplicate rows:", int(raw_df.duplicated().sum()))
print("Duplicate timestamps:", int(ts.duplicated().sum()))
```

```python
data_cfg = DataConfig()
target_cfg = TargetConfig(horizon=6, threshold=0.0075)
split_cfg = SplitConfig(train_ratio=0.70, val_ratio=0.15, test_ratio=0.15)

metadata = prepare_datasets(
    data_config=data_cfg,
    target_config=target_cfg,
    split_config=split_cfg,
    corr_threshold=0.95,
    apply_vif=True,
    vif_threshold=10.0,
)
metadata
```
